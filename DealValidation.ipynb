```python
# Extract text from PDF
!pip install PyPDF2

import PyPDF2

def extract_text_from_pdf(pdf_path):
  with open(pdf_path, 'rb') as pdf_file:
    pdf_reader = PyPDF2.PdfReader(pdf_file)
    text = ''
    for page_num in range(len(pdf_reader.pages)):
      page = pdf_reader.pages[page_num]
      text += page.extract_text()
    return text

# Example usage
pdf_file_path = '/content/sample_data/DSA Roadmap.pdf' # replace with the actual path to your PDF file
extracted_text = extract_text_from_pdf(pdf_file_path)
extracted_text
```

```python
# Extract text from Excel
!pip install openpyxl

import openpyxl

def extract_data_from_excel(excel_file_path):
    try:
        workbook = openpyxl.load_workbook(excel_file_path, data_only=True) # data_only=True to get calculated values
        sheet = workbook.active  # Get the active sheet

        data = []
        for row in sheet.iter_rows():
            row_data = [cell.value for cell in row]
            data.append(row_data)
        return data
    except FileNotFoundError:
        print(f"Error: File not found at {excel_file_path}")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None

# Example usage:
excel_file_path = '/content/mySample_data/Sample Deals 1.xlsx' # replace with the actual path
extracted_data = extract_data_from_excel(excel_file_path)
extracted_data
```

```python
# prompt: extract image and text from excel

!pip install openpyxl Pillow

import openpyxl
from PIL import Image
from io import BytesIO

def extract_image_from_excel(excel_file_path, image_index=0):
    """
    Extracts an image from an Excel file.

    Args:
        excel_file_path: The path to the Excel file.
        image_index: The index of the image to extract (default is 0 for the first image).

    Returns:
        A PIL Image object if an image is found, None otherwise.
    """
    try:
        workbook = openpyxl.load_workbook(excel_file_path, data_only=True)
        sheet = workbook.active

        image_count = 0
        for row in sheet.iter_rows():
            for cell in row:
                if cell.drawing is not None:
                    for drawing in cell.drawing._drawing.anchors:
                        if image_count == image_index:
                            image = drawing.graphic.graphicData.pic.blipFill.blip
                            image_data = image.embed
                            image_bytes = workbook._images[image_data]  # Retrieve image bytes
                            image = Image.open(BytesIO(image_bytes))
                            return image
                        image_count += 1
        print("No image found at the specified index.")
        return None

    except FileNotFoundError:
        print(f"Error: File not found at {excel_file_path}")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None

# Example usage
excel_file_path = '//content/mySample_data/Sample Deals 1.xlsx'  # Replace with the actual path
extracted_image = extract_image_from_excel(excel_file_path)

if extracted_image:
    # Display the image (optional)
    extracted_image.show()

    # Save the image (optional)
    extracted_image.save("extracted_image.png")
```

```python
# Extract text from PPT
!pip install python-pptx

from pptx import Presentation

def extract_text_from_ppt(ppt_path):
    prs = Presentation(ppt_path)
    text = ""
    for slide in prs.slides:
        for shape in slide.shapes:
            if hasattr(shape, "text"):
                text += shape.text + "\n"
    return text

# Example usage
ppt_file_path = '/content/sample_data/samplepptx.pptx'  # Replace with the actual path to your PPT file
extracted_text = extract_text_from_ppt(ppt_file_path)
extracted_text
```

```python
# Extract text from Doc
!pip install python-docx

import docx

def extract_text_from_docx(docx_path):
    doc = docx.Document(docx_path)
    full_text = []
    for para in doc.paragraphs:
        full_text.append(para.text)
    return '\n'.join(full_text)

# Example usage
docx_file_path = '/content/sample_data/Mohammad Ahsan_Word (1).docx'  # Replace with the actual path to your .docx file
extracted_text = extract_text_from_docx(docx_file_path)
extracted_text
```

```python
import os

def extract_text_from_file(file_path):
    """Extracts text from various file types (PDF, Excel, PPT, DOCX)."""
    try:
        _, file_extension = os.path.splitext(file_path)
        file_extension = file_extension.lower()

        if file_extension == ".pdf":
            # Code to extract text from PDF (existing code can be reused here)
            import PyPDF2
            with open(file_path, 'rb') as pdf_file:
                pdf_reader = PyPDF2.PdfReader(pdf_file)
                text = ''
                for page_num in range(len(pdf_reader.pages)):
                    page = pdf_reader.pages[page_num]
                    text += page.extract_text()
            return text
        elif file_extension == ".xlsx":
            import openpyxl
            workbook = openpyxl.load_workbook(file_path, data_only=True)
            sheet = workbook.active
            data = []
            for row in sheet.iter_rows():
                row_data = [cell.value for cell in row]
                data.append(row_data)
            return str(data)  # Convert to string for consistent output
        elif file_extension == ".pptx":
            from pptx import Presentation
            prs = Presentation(file_path)
            text = ""
            for slide in prs.slides:
                for shape in slide.shapes:
                    if hasattr(shape, "text"):
                        text += shape.text + "\n"
            return text
        elif file_extension == ".docx":
            import docx
            doc = docx.Document(file_path)
            full_text = []
            for para in doc.paragraphs:
                full_text.append(para.text)
            return '\n'.join(full_text)
        elif file_extension == ".txt":
            with open(file_path, 'r') as f:
                return f.read()
        else:
            return f"Unsupported file type: {file_extension}"

    except FileNotFoundError:
        return f"Error: File not found at {file_path}"
    except Exception as e:
        return f"An error occurred: {e}"

# Example usage (replace with your file path)
file_path = '/content/sample_data/samplepptx.pptx'
extracted_content = extract_text_from_file(file_path)
extracted_content
```
