```python
# Extract text from Excel
!pip install openpyxl

import openpyxl

def extract_data_from_excel(excel_file_path):
    try:
        workbook = openpyxl.load_workbook(excel_file_path, data_only=True) # data_only=True to get calculated values
        sheet = workbook.active  # Get the active sheet

        data = []
        for row in sheet.iter_rows():
            row_data = [cell.value for cell in row]
            data.append(row_data)
        return data
    except FileNotFoundError:
        print(f"Error: File not found at {excel_file_path}")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None

# Example usage:
excel_file_path = '/content/Sample Deals 1.xlsx' # replace with the actual path
extracted_data = extract_data_from_excel(excel_file_path)
extracted_data
```

```python
# prompt: extract text from PDF

import PyPDF2

def extract_text_from_pdf(pdf_path):
  try:
    with open(pdf_path, 'rb') as pdf_file:
      pdf_reader = PyPDF2.PdfReader(pdf_file)
      text = ''
      for page_num in range(len(pdf_reader.pages)):
        page = pdf_reader.pages[page_num]
        text += page.extract_text()
      return text
  except FileNotFoundError:
    print(f"Error: File not found at {pdf_path}")
    return None
  except PyPDF2.errors.PdfReadError:
    print(f"Error: Could not read PDF file at {pdf_path}.  It may be corrupted or encrypted.")
    return None
  except Exception as e:
    print(f"An unexpected error occurred: {e}")
    return None
# Example usage (replace with your file path)
pdf_file_path = '/content/Mohammad Ahsan_NTTD_Resume (1).pdf'
extracted_text = extract_text_from_pdf(pdf_file_path)

if extracted_text: extracted_text

print(extracted_text)
```

```python
# Extract text from PPT
!pip install python-pptx

from pptx import Presentation

def extract_text_from_ppt(ppt_path):
    prs = Presentation(ppt_path)
    text = ""
    for slide in prs.slides:
        for shape in slide.shapes:
            if hasattr(shape, "text"):
                text += shape.text + "\n"
    return text

# Example usage
ppt_file_path = '/content/Gainwell VRB.pptx'  # Replace with the actual path to your PPT file
extracted_text = extract_text_from_ppt(ppt_file_path)
extracted_text
```

```python
# Extract text from Doc
!pip install python-docx

import docx

def extract_text_from_docx(docx_path):
    doc = docx.Document(docx_path)
    full_text = []
    for para in doc.paragraphs:
        full_text.append(para.text)
    return '\n'.join(full_text)

# Example usage
docx_file_path = '/content/Mohammad Ahsan_NTTD_Resume.docx'  # Replace with the actual path to your .docx file
extracted_text = extract_text_from_docx(docx_file_path)
extracted_text
```

```python
import os

def extract_text_from_file(file_path):
    """Extracts text from various file types (PDF, Excel, PPT, DOCX)."""
    try:
        _, file_extension = os.path.splitext(file_path)
        file_extension = file_extension.lower()

        if file_extension == ".pdf":
            # Code to extract text from PDF (existing code can be reused here)
            import PyPDF2
            with open(file_path, 'rb') as pdf_file:
                pdf_reader = PyPDF2.PdfReader(pdf_file)
                text = ''
                for page_num in range(len(pdf_reader.pages)):
                    page = pdf_reader.pages[page_num]
                    text += page.extract_text()
            return text
        elif file_extension == ".xlsx":
            import openpyxl
            workbook = openpyxl.load_workbook(file_path, data_only=True)
            sheet = workbook.active
            data = []
            for row in sheet.iter_rows():
                row_data = [cell.value for cell in row]
                data.append(row_data)
            return str(data)  # Convert to string for consistent output
        elif file_extension == ".pptx":
            from pptx import Presentation
            prs = Presentation(file_path)
            text = ""
            for slide in prs.slides:
                for shape in slide.shapes:
                    if hasattr(shape, "text"):
                        text += shape.text + "\n"
            return text
        elif file_extension == ".docx":
            import docx
            doc = docx.Document(file_path)
            full_text = []
            for para in doc.paragraphs:
                full_text.append(para.text)
            return '\n'.join(full_text)
        elif file_extension == ".txt":
            with open(file_path, 'r') as f:
                return f.read()
        else:
            return f"Unsupported file type: {file_extension}"

    except FileNotFoundError:
        return f"Error: File not found at {file_path}"
    except Exception as e:
        return f"An error occurred: {e}"

# Example usage (replace with your file path)
file_path = '/content/sample_data/samplepptx.pptx'
extracted_content = extract_text_from_file(file_path)
extracted_content
```

```python
#extract data from xlsb using pyxlsb

!pip install pyxlsb

import pyxlsb

def extract_data_from_xlsb(xlsb_file_path):
    try:
        data = []
        with pyxlsb.open_workbook(xlsb_file_path) as wb:
            for sheet_name in wb.sheets:
                with wb.get_sheet(sheet_name) as sheet:
                    for row in sheet.rows():
                        data.append([item.v for item in row])  # Extract cell values
        return data
    except FileNotFoundError:
        print(f"Error: File not found at {xlsb_file_path}")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None

# Example usage
xlsb_file_path = '/content/Gainwell_Data Center_0198179_2025.01.02_vFinal 1.0_POS.xlsb'  # Replace with the actual path to your .xlsb file
extracted_data = extract_data_from_xlsb(xlsb_file_path)

if extracted_data: extracted_data
print(extracted_data)
```

```python
# prompt: get data from specific sheet of excel workbook and save the data in txt file

import openpyxl

def extract_sheet_data(excel_file, sheet_name, output_file):
    try:
        workbook = openpyxl.load_workbook(excel_file, data_only=True)
        if sheet_name not in workbook.sheetnames:
            print(f"Error: Sheet '{sheet_name}' not found in the workbook.")
            return

        sheet = workbook[sheet_name]
        with open(output_file, 'w') as f:
            for row in sheet.iter_rows():
                row_data = [str(cell.value) if cell.value is not None else '' for cell in row]
                f.write('\t'.join(row_data) + '\n')  # Join with tabs for better readability

        print(f"Data from sheet '{sheet_name}' extracted and saved to '{output_file}'.")

    except FileNotFoundError:
        print(f"Error: File not found at {excel_file}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage:
excel_file_path = '/content/XLSampSheets.xlsx'  # Replace with your Excel file path
sheet_to_extract = 'Sheet5'  # Replace with the actual sheet name
output_txt_file = '/content/ext.txt'  # Replace with the desired output file path

extract_sheet_data(excel_file_path, sheet_to_extract, output_txt_file)
```

```python
# prompt: fetch data from ppt and store it in txt file

def extract_ppt_to_txt(ppt_path, txt_path):
    try:
        prs = Presentation(ppt_path)
        with open(txt_path, "w") as txt_file:
            for slide in prs.slides:
                for shape in slide.shapes:
                    if hasattr(shape, "text"):
                        txt_file.write(shape.text + "\n")
        print(f"Data from '{ppt_path}' extracted and saved to '{txt_path}'")
    except FileNotFoundError:
        print(f"Error: File not found at {ppt_path}")
    except Exception as e:
        print(f"An error occurred: {e}")


# Example usage
ppt_file_path = '/content/Gainwell VRB.pptx'  # Replace with the actual path to your PPT file
output_txt_file = '/content/extPPT.txt'  # Replace with the desired output file path
extract_ppt_to_txt(ppt_file_path, output_txt_file)
```
