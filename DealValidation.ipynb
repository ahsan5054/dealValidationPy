```python
# prompt: extract text data from PDF and save it in text file

!pip install PyPDF2

import PyPDF2

def extract_text_from_pdf(pdf_path, text_path):
    with open(pdf_path, 'rb') as pdf_file, open(text_path, 'w') as text_file:
        pdf_reader = PyPDF2.PdfReader(pdf_file)
        for page_num in range(len(pdf_reader.pages)):
            page = pdf_reader.pages[page_num]
            text = page.extract_text()
            text_file.write(text)

# Example usage:
file_path = '/content/50pagesample'
pdf_file_path = file_path+'.pdf' # Replace with the actual path to your PDF file
text_file_path = file_path+'Text'+'.txt' # Replace with the desired output text file path
extract_text_from_pdf(pdf_file_path, text_file_path)
```

```python
# prompt: get data from specific sheet of excel workbook and save the data in txt file

import openpyxl

def extract_sheet_data(excel_file, sheet_name, output_file):
    try:
        workbook = openpyxl.load_workbook(excel_file, data_only=True)
        if sheet_name not in workbook.sheetnames:
            print(f"Error: Sheet '{sheet_name}' not found in the workbook.")
            return

        sheet = workbook[sheet_name]
        with open(output_file, 'w') as f:
            for row in sheet.iter_rows():
                row_data = [str(cell.value) if cell.value is not None else '' for cell in row]
                f.write('\t'.join(row_data) + '\n')  # Join with tabs for better readability

        print(f"Data from sheet '{sheet_name}' extracted and saved to '{output_file}'.")

    except FileNotFoundError:
        print(f"Error: File not found at {excel_file}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage:
file_path = '/content/Worksheet in Sample Deals 1'
excel_file_path = file_path+'.xlsx'  # Replace with your Excel file path
sheet_to_extract = 'Summary'  # Replace with the actual sheet name
output_txt_file = file_path+'Text'+'.txt'  # Replace with the desired output file path

extract_sheet_data(excel_file_path, sheet_to_extract, output_txt_file)
```

```python
# prompt: fetch data from ppt and store it in txt file
!pip install python-pptx
from pptx import Presentation
def extract_ppt_to_txt(ppt_path, txt_path):
    try:
        prs = Presentation(ppt_path)
        with open(txt_path, "w") as txt_file:
            for slide in prs.slides:
                for shape in slide.shapes:
                    if hasattr(shape, "text"):
                        txt_file.write(shape.text + "\n")
        print(f"Data from '{ppt_path}' extracted and saved to '{txt_path}'")
    except FileNotFoundError:
        print(f"Error: File not found at {ppt_path}")
    except Exception as e:
        print(f"An error occurred: {e}")


# Example usage
file_path = '/content/Gainwell VRB'
ppt_file_path = file_path+'.pptx'  # Replace with the actual path to your PPT file
output_txt_file = file_path+'Text'+'.txt'  # Replace with the desired output file path
extract_ppt_to_txt(ppt_file_path, output_txt_file)
```

```python
# prompt: Fetch text data from doc file and store it in text file

!pip install python-docx

import docx

def extract_text_from_doc(doc_path, text_path):
    try:
        doc = docx.Document(doc_path)
        with open(text_path, 'w', encoding='utf-8') as text_file:
            for paragraph in doc.paragraphs:
                text_file.write(paragraph.text + '\n')
        print(f"Text extracted from '{doc_path}' and saved to '{text_path}'")
    except FileNotFoundError:
        print(f"Error: File not found at {doc_path}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
file_path = '/content/demo'
doc_file_path = file_path+'.docx'  # Replace with the actual path to your DOCX file
output_text_file = file_path+'Text'+'.txt'  # Replace with the desired output file path
extract_text_from_doc(doc_file_path, output_text_file)
```

```python
# prompt: Fetch text from xlsb file and save in txt file

!pip install pyxlsb

import pyxlsb
import os

def extract_xlsb_text(xlsb_path, txt_path):
    try:
        with open(txt_path, 'w', encoding='utf-8') as txt_file:
            wb = pyxlsb.open_workbook(xlsb_path)
            for sheet_name in wb.sheets:
                with wb.get_sheet(sheet_name) as sheet:
                    for row in sheet.rows():
                        for cell in row:
                            if cell.v is not None:  # Check if the cell has a value
                                txt_file.write(str(cell.v) + '\t') # Write cell value followed by tab
                        txt_file.write('\n') # New line after each row
            print(f"Data extracted from '{xlsb_path}' and saved to '{txt_path}'")

    except FileNotFoundError:
        print(f"Error: File not found at {xlsb_path}")
    except Exception as e:
        print(f"An error occurred: {e}")


# Example usage:
file_path = '/content/Gainwell_Data Center' # Replace with the actual path to your xlsb file
xlsb_file_path = file_path + '.xlsb'
output_txt_file = file_path + 'Text.txt'

extract_xlsb_text(xlsb_file_path, output_txt_file)
```

```python
# prompt: In amazon s3 bucket when ever a .pptx or .xlxs file is loaded fetch all the text data to create and new text file and upload in same folder

import boto3
import os
from io import BytesIO
from pptx import Presentation
import openpyxl
import pyxlsb
import docx

# Replace with your actual AWS credentials and S3 bucket details
AWS_ACCESS_KEY_ID = 'YOUR_ACCESS_KEY'
AWS_SECRET_ACCESS_KEY = 'YOUR_SECRET_KEY'
S3_BUCKET_NAME = 'your_bucket_name'

s3 = boto3.client('s3', aws_access_key_id=AWS_ACCESS_KEY_ID, aws_secret_access_key=AWS_SECRET_ACCESS_KEY)

def extract_text_from_pptx(pptx_file):
    prs = Presentation(pptx_file)
    text = ""
    for slide in prs.slides:
        for shape in slide.shapes:
            if hasattr(shape, "text"):
                text += shape.text + "\n"
    return text

def extract_text_from_xlsx(xlsx_file, sheet_name):
  try:
    workbook = openpyxl.load_workbook(xlsx_file, data_only=True)
    if sheet_name not in workbook.sheetnames:
      print(f"Error: Sheet '{sheet_name}' not found in the workbook.")
      return ""
    sheet = workbook[sheet_name]
    text = ""
    for row in sheet.iter_rows():
        for cell in row:
            if cell.value is not None:
                text += str(cell.value) + "\t"
        text += "\n"
    return text
  except Exception as e:
      print(f"Error processing xlsx file: {e}")
      return ""


def extract_text_from_xlsb(xlsb_file):
    try:
      text = ""
      wb = pyxlsb.open_workbook(xlsb_file)
      for sheet_name in wb.sheets:
        with wb.get_sheet(sheet_name) as sheet:
            for row in sheet.rows():
                for cell in row:
                    if cell.v is not None:
                        text += str(cell.v) + "\t"
                text += "\n"
      return text
    except Exception as e:
      print(f"Error processing xlsb file: {e}")
      return ""


def lambda_handler(event, context):
    bucket_name = event['Records'][0]['s3']['bucket']['name']
    key = event['Records'][0]['s3']['object']['key']

    if key.lower().endswith(('.pptx', '.xlsx', '.xlsb')):
        try:
            obj = s3.get_object(Bucket=bucket_name, Key=key)
            file_content = obj['Body'].read()
            file_stream = BytesIO(file_content)

            if key.lower().endswith('.pptx'):
              text_data = extract_text_from_pptx(file_stream)
            elif key.lower().endswith('.xlsx'):
              text_data = extract_text_from_xlsx(file_stream, "Sheet1") # Replace "Sheet1" with actual sheet name if necessary
            elif key.lower().endswith('.xlsb'):
              text_data = extract_text_from_xlsb(file_stream)

            # Create output file name and upload to S3
            output_key = os.path.splitext(key)[0] +'Text' +'.txt'
            s3.put_object(Bucket=bucket_name, Key=output_key, Body=text_data)
            print(f"Text file '{output_key}' created and uploaded to S3.")

        except Exception as e:
            print(f"Error processing file: {e}")
            return {
                'statusCode': 500,
                'body': f'Error processing file: {str(e)}'
            }
    else:
        print(f"Skipping file {key}, not a .pptx, .xlsx, or .xlsb file.")

    return {
        'statusCode': 200,
        'body': 'File processed successfully!'
    }
```
