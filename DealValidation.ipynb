```python
# prompt: extract text data from PDF and save it in text file

!pip install PyPDF2

import PyPDF2

def extract_text_from_pdf(pdf_path, text_path):
    with open(pdf_path, 'rb') as pdf_file, open(text_path, 'w') as text_file:
        pdf_reader = PyPDF2.PdfReader(pdf_file)
        for page_num in range(len(pdf_reader.pages)):
            page = pdf_reader.pages[page_num]
            text = page.extract_text()
            text_file.write(text)

# Example usage:
file_path = '/content/50pagesample'
pdf_file_path = file_path+'.pdf' # Replace with the actual path to your PDF file
text_file_path = file_path+'Text'+'.txt' # Replace with the desired output text file path
extract_text_from_pdf(pdf_file_path, text_file_path)
```

```python
# prompt: get data from specific sheet of excel workbook and save the data in txt file

import openpyxl

def extract_sheet_data(excel_file, sheet_name, output_file):
    try:
        workbook = openpyxl.load_workbook(excel_file, data_only=True)
        if sheet_name not in workbook.sheetnames:
            print(f"Error: Sheet '{sheet_name}' not found in the workbook.")
            return

        sheet = workbook[sheet_name]
        with open(output_file, 'w') as f:
            for row in sheet.iter_rows():
                row_data = [str(cell.value) if cell.value is not None else '' for cell in row]
                f.write('\t'.join(row_data) + '\n')  # Join with tabs for better readability

        print(f"Data from sheet '{sheet_name}' extracted and saved to '{output_file}'.")

    except FileNotFoundError:
        print(f"Error: File not found at {excel_file}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage:
file_path = '/content/excel_files/Worksheet in Sample Deals 1'
excel_file_path = file_path+'.xlsx'  # Replace with your Excel file path
sheet_to_extract = 'Summary'  # Replace with the actual sheet name
output_txt_file = file_path+'Text'+'.txt'  # Replace with the desired output file path

extract_sheet_data(excel_file_path, sheet_to_extract, output_txt_file)
```

```python
# prompt: fetch data from ppt and store it in txt file
!pip install python-pptx
from pptx import Presentation
def extract_ppt_to_txt(ppt_path, txt_path):
    try:
        prs = Presentation(ppt_path)
        with open(txt_path, "w") as txt_file:
            for slide in prs.slides:
                for shape in slide.shapes:
                    if hasattr(shape, "text"):
                        txt_file.write(shape.text + "\n")
        print(f"Data from '{ppt_path}' extracted and saved to '{txt_path}'")
    except FileNotFoundError:
        print(f"Error: File not found at {ppt_path}")
    except Exception as e:
        print(f"An error occurred: {e}")


# Example usage
file_path = '/content/Gainwell VRB'
ppt_file_path = file_path+'.pptx'  # Replace with the actual path to your PPT file
output_txt_file = file_path+'Text'+'.txt'  # Replace with the desired output file path
extract_ppt_to_txt(ppt_file_path, output_txt_file)
```

```python
# prompt: Fetch text data from doc file and store it in text file

!pip install python-docx

import docx

def extract_text_from_doc(doc_path, text_path):
    try:
        doc = docx.Document(doc_path)
        with open(text_path, 'w', encoding='utf-8') as text_file:
            for paragraph in doc.paragraphs:
                text_file.write(paragraph.text + '\n')
        print(f"Text extracted from '{doc_path}' and saved to '{text_path}'")
    except FileNotFoundError:
        print(f"Error: File not found at {doc_path}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
file_path = '/content/demo'
doc_file_path = file_path+'.docx'  # Replace with the actual path to your DOCX file
output_text_file = file_path+'Text'+'.txt'  # Replace with the desired output file path
extract_text_from_doc(doc_file_path, output_text_file)
```

```python
# prompt: Fetch text from xlsb file and save in txt file

!pip install pyxlsb

import pyxlsb
import os

def extract_xlsb_text(xlsb_path, txt_path):
    try:
        with open(txt_path, 'w', encoding='utf-8') as txt_file:
            wb = pyxlsb.open_workbook(xlsb_path)
            for sheet_name in wb.sheets:
                with wb.get_sheet(sheet_name) as sheet:
                    for row in sheet.rows():
                        for cell in row:
                            if cell.v is not None:  # Check if the cell has a value
                                txt_file.write(str(cell.v) + '\t') # Write cell value followed by tab
                        txt_file.write('\n') # New line after each row
            print(f"Data extracted from '{xlsb_path}' and saved to '{txt_path}'")

    except FileNotFoundError:
        print(f"Error: File not found at {xlsb_path}")
    except Exception as e:
        print(f"An error occurred: {e}")


# Example usage:
file_path = '/content/Gainwell_Data Center' # Replace with the actual path to your xlsb file
xlsb_file_path = file_path + '.xlsb'
output_txt_file = file_path + 'Text.txt'

extract_xlsb_text(xlsb_file_path, output_txt_file)
```

```python
# prompt: In amazon s3 bucket when ever a .pptx or .xlxs file is loaded fetch all the text data to create and new text file and upload in same folder

import boto3
import os
from io import BytesIO
from pptx import Presentation
import openpyxl
import pyxlsb
import docx

# Replace with your actual AWS credentials and S3 bucket details
AWS_ACCESS_KEY_ID = 'YOUR_ACCESS_KEY'
AWS_SECRET_ACCESS_KEY = 'YOUR_SECRET_KEY'
S3_BUCKET_NAME = 'your_bucket_name'

s3 = boto3.client('s3', aws_access_key_id=AWS_ACCESS_KEY_ID, aws_secret_access_key=AWS_SECRET_ACCESS_KEY)

def extract_text_from_pptx(pptx_file):
    prs = Presentation(pptx_file)
    text = ""
    for slide in prs.slides:
        for shape in slide.shapes:
            if hasattr(shape, "text"):
                text += shape.text + "\n"
    return text

def extract_text_from_xlsx(xlsx_file, sheet_name):
  try:
    workbook = openpyxl.load_workbook(xlsx_file, data_only=True)
    if sheet_name not in workbook.sheetnames:
      print(f"Error: Sheet '{sheet_name}' not found in the workbook.")
      return ""
    sheet = workbook[sheet_name]
    text = ""
    for row in sheet.iter_rows():
        for cell in row:
            if cell.value is not None:
                text += str(cell.value) + "\t"
        text += "\n"
    return text
  except Exception as e:
      print(f"Error processing xlsx file: {e}")
      return ""


def extract_text_from_xlsb(xlsb_file):
    try:
      text = ""
      wb = pyxlsb.open_workbook(xlsb_file)
      for sheet_name in wb.sheets:
        with wb.get_sheet(sheet_name) as sheet:
            for row in sheet.rows():
                for cell in row:
                    if cell.v is not None:
                        text += str(cell.v) + "\t"
                text += "\n"
      return text
    except Exception as e:
      print(f"Error processing xlsb file: {e}")
      return ""


def lambda_handler(event, context):
    bucket_name = event['Records'][0]['s3']['bucket']['name']
    key = event['Records'][0]['s3']['object']['key']

    if key.lower().endswith(('.pptx', '.xlsx', '.xlsb')):
        try:
            obj = s3.get_object(Bucket=bucket_name, Key=key)
            file_content = obj['Body'].read()
            file_stream = BytesIO(file_content)

            if key.lower().endswith('.pptx'):
              text_data = extract_text_from_pptx(file_stream)
            elif key.lower().endswith('.xlsx'):
              text_data = extract_text_from_xlsx(file_stream, "Sheet1") # Replace "Sheet1" with actual sheet name if necessary
            elif key.lower().endswith('.xlsb'):
              text_data = extract_text_from_xlsb(file_stream)

            # Create output file name and upload to S3
            output_key = os.path.splitext(key)[0] +'Text' +'.txt'
            s3.put_object(Bucket=bucket_name, Key=output_key, Body=text_data)
            print(f"Text file '{output_key}' created and uploaded to S3.")

        except Exception as e:
            print(f"Error processing file: {e}")
            return {
                'statusCode': 500,
                'body': f'Error processing file: {str(e)}'
            }
    else:
        print(f"Skipping file {key}, not a .pptx, .xlsx, or .xlsb file.")

    return {
        'statusCode': 200,
        'body': 'File processed successfully!'
    }
```

```python
# prompt: read data from all the image in the folder. Create and save text file of details extracted from all the image in a single text file.

import os
from PIL import Image
import pytesseract

def extract_image_text(image_path, output_file):
    try:
        img = Image.open(image_path)
        text = pytesseract.image_to_string(img)
        with open(output_file, "a") as f:
            f.write(f"Image: {image_path}\n")
            f.write(text + "\n\n") # Add separator between images
        print(f"Text from {image_path} appended to {output_file}")
    except Exception as e:
        print(f"Error processing {image_path}: {e}")

image_folder = "/content/images" # Replace with actual folder path
output_text_file = "/content/images/image_text.txt" # Replace with desired output path

# Ensure output file exists and is empty
open(output_text_file, 'w').close()

for filename in os.listdir(image_folder):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif')):
        image_path = os.path.join(image_folder, filename)
        extract_image_text(image_path, output_text_file)
```

```python
# prompt: from multiple excel workbook fetch text data from the specific sheet of the excel which has the matching keyword. Create and save text data in text file in same folder

import os
import openpyxl
import re

def extract_text_from_excel_files(folder_path, keyword, output_file):
    """
    Extracts text data from specific sheets in multiple Excel workbooks within a folder.

    Args:
        folder_path: The path to the folder containing the Excel files.
        keyword: The keyword to search for in sheet names.
        output_file: The path to the output text file.
    """

    try:
        with open(output_file, 'w', encoding='utf-8') as outfile:
          for filename in os.listdir(folder_path):
              if filename.endswith(('.xlsx', '.xlsm', '.xltx', '.xltm')): # Include other Excel formats if needed
                  filepath = os.path.join(folder_path, filename)
                  try:
                      workbook = openpyxl.load_workbook(filepath, data_only=True)
                      for sheet_name in workbook.sheetnames:
                          if re.search(keyword, sheet_name, re.IGNORECASE): # Use re.search for flexible keyword matching
                              sheet = workbook[sheet_name]
                              for row in sheet.iter_rows():
                                  row_data = [str(cell.value) if cell.value is not None else '' for cell in row]
                                  outfile.write('\t'.join(row_data) + '\n')
                              print(f"Data from sheet '{sheet_name}' in '{filename}' extracted.")
                  except Exception as e:
                      print(f"Error processing {filename}: {e}")

        print(f"Text data saved to {output_file}")

    except FileNotFoundError:
        print(f"Error: Folder not found at {folder_path}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example Usage
folder_path = '/content/excel_files'  # Replace with the actual folder path
keyword = 'sheet1'  # Replace with your desired keyword
output_file = '/content/excel_files/extracted_text.txt'  # Replace with the desired output file path

extract_text_from_excel_files(folder_path, keyword, output_file)
```

```python
# prompt: from the excel file read data from specific sheet and create human readable text file and save in same folder

def extract_sheet_data_and_save(excel_file, sheet_name, output_folder):
    """Reads data from a specific sheet in an Excel file and saves it as a human-readable text file.

    Args:
        excel_file: Path to the Excel file.
        sheet_name: Name of the sheet to extract data from.
        output_folder: Folder to save the text file.
    """
    try:
        workbook = openpyxl.load_workbook(excel_file, data_only=True)
        if sheet_name not in workbook.sheetnames:
            print(f"Error: Sheet '{sheet_name}' not found in the workbook.")
            return

        sheet = workbook[sheet_name]
        output_file_path = os.path.join(output_folder, f"{sheet_name}_extracted.txt")

        with open(output_file_path, 'w', encoding='utf-8') as f:
            for row in sheet.iter_rows():
                row_data = [str(cell.value) if cell.value is not None else '' for cell in row]
                f.write('\t'.join(row_data) + '\n')

        print(f"Data from sheet '{sheet_name}' extracted and saved to '{output_file_path}'.")

    except FileNotFoundError:
        print(f"Error: File not found at {excel_file}")
    except Exception as e:
        print(f"An error occurred: {e}")


# Example usage
excel_file_path = '/content/excel_files/Worksheet in Sample Deals 1.xlsx'  # Replace with the actual path to your Excel file
sheet_to_extract = 'Summary'  # Replace with the actual sheet name
output_folder = os.path.dirname(excel_file_path)  # Save in the same folder as Excel file

extract_sheet_data_and_save(excel_file_path, sheet_to_extract, output_folder)
```

```python
# prompt: from excel sheet convert specific sheet in to pdf in all data in not coming rotate the pdf sheet and get all data. All the data from sheet in not coming in pdf. Adjust everything on the sheet to pdf. No data should be lost. optimize below code

import openpyxl
import os
from pathlib import Path
from io import BytesIO
from pptx import Presentation
import pyxlsb
import re

!pip install reportlab

from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas

def excel_to_pdf(excel_file, sheet_name, pdf_file):
    try:
        workbook = openpyxl.load_workbook(excel_file, data_only=True)
        if sheet_name not in workbook.sheetnames:
            print(f"Error: Sheet '{sheet_name}' not found in the workbook.")
            return

        sheet = workbook[sheet_name]
        c = canvas.Canvas(pdf_file, pagesize=letter)
        width, height = letter
        x_offset = 50
        y_offset = height - 50
        line_height = 15

        for row in sheet.iter_rows():
            for cell in row:
                if cell.value is not None:
                  text = str(cell.value)
                  c.drawString(x_offset, y_offset, text)
                  x_offset += 100 # Adjust as needed

            y_offset -= line_height
            x_offset = 50 # Reset x for new line
            if y_offset < 50:
              c.showPage()
              y_offset = height - 50

        c.save()
        print(f"Sheet '{sheet_name}' from '{excel_file}' converted to '{pdf_file}'")

    except FileNotFoundError:
        print(f"Error: File not found at {excel_file}")
    except Exception as e:
        print(f"An error occurred: {e}")


# Example usage:
excel_file_path = '/content/excel_files/Worksheet in Sample Deals 1.xlsx'  # Replace with your Excel file path
sheet_to_extract = 'Summary'  # Replace with the actual sheet name
pdf_file_path = '/content/excel_files/summary.pdf'

excel_to_pdf(excel_file_path, sheet_to_extract, pdf_file_path)


def extract_text_from_excel_files(folder_path, keyword, output_file):
    """
    Extracts text data from specific sheets in multiple Excel workbooks within a folder.

    Args:
        folder_path: The path to the folder containing the Excel files.
        keyword: The keyword to search for in sheet names.
        output_file: The path to the output text file.
    """
    try:
        with open(output_file, 'w', encoding='utf-8') as outfile:
            for filename in os.listdir(folder_path):
                if filename.endswith(('.xlsx', '.xlsm', '.xltx', '.xltm', '.xlsb')):
                    filepath = os.path.join(folder_path, filename)
                    try:
                        if filename.endswith('.xlsb'):
                            wb = pyxlsb.open_workbook(filepath)
                        else:
                            wb = openpyxl.load_workbook(filepath, data_only=True)

                        for sheet_name in wb.sheetnames:
                            if re.search(keyword, sheet_name, re.IGNORECASE):
                                if filename.endswith('.xlsb'):
                                    with wb.get_sheet(sheet_name) as sheet:
                                        for row in sheet.rows():
                                            row_data = [str(cell.v) if cell.v is not None else '' for cell in row]
                                            outfile.write('\t'.join(row_data) + '\n')
                                else:
                                    sheet = wb[sheet_name]
                                    for row in sheet.iter_rows():
                                        row_data = [str(cell.value) if cell.value is not None else '' for cell in row]
                                        outfile.write('\t'.join(row_data) + '\n')
                                print(f"Data from sheet '{sheet_name}' in '{filename}' extracted.")
                    except Exception as e:
                        print(f"Error processing {filename}: {e}")

        print(f"Text data saved to {output_file}")
    except FileNotFoundError:
        print(f"Error: Folder not found at {folder_path}")
    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
# prompt: from a folder of multiple excel file convert to single merged text file with each sheet data. Keep 'Excel name-sheet name' as heading of each sheet.

import os
import openpyxl
import re

def merge_excel_to_text(excel_folder, output_file):
    """
    Merges data from multiple Excel files in a folder into a single text file.

    Args:
        excel_folder: Path to the folder containing Excel files.
        output_file: Path to the output text file.
    """
    try:
        with open(output_file, 'w', encoding='utf-8') as outfile:
            for filename in os.listdir(excel_folder):
                if filename.endswith(('.xlsx', '.xlsm', '.xltx', '.xltm')):
                    filepath = os.path.join(excel_folder, filename)
                    try:
                        workbook = openpyxl.load_workbook(filepath, data_only=True)
                        for sheet_name in workbook.sheetnames:
                            sheet = workbook[sheet_name]
                            outfile.write(f"Excel Name: {filename}, Sheet Name: {sheet_name}\n")  # Header
                            for row in sheet.iter_rows():
                                row_data = [str(cell.value) if cell.value is not None else '' for cell in row]
                                outfile.write('\t'.join(row_data) + '\n')
                            outfile.write("\n")  # Separator between sheets
                    except Exception as e:
                        print(f"Error processing {filename}: {e}")
        print(f"Merged data saved to {output_file}")

    except FileNotFoundError:
        print(f"Error: Folder not found at {excel_folder}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example Usage
excel_folder_path = '/content/excels'  # Replace with your folder path
output_text_file = '/content/excels/merged_excel_data.txt'  # Replace with your desired output file path

merge_excel_to_text(excel_folder_path, output_text_file)
```
